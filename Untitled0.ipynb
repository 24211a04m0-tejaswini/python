```python
# ── Simple if-else ────────────────────────────────────────────────────
number = -7   # the number to test

if number > 0:              # condition: is number positive?
    print("Positive number")
elif number == 0:           # else-if: is number zero?
    print("Number is zero")
else:                       # none of the above matched
    print("Negative number")
```

```python
# ── Piecewise mathematical function ──────────────────────────────────
# f(x) = x²  if x >= 0
#        -x  if x < 0

def f(x):
    if x >= 0:         # non-negative: use x squared
        return x ** 2
    else:              # negative: use absolute value
        return -x

# Test the function for different values
for val in [-3, -1, 0, 2, 5]:
    print(f"f({val:3d}) = {f(val)}")
```

```python
# ── Grade classification using if-elif-else ──────────────────────────
score = 73   # student's exam score

if score >= 90:
    grade = "A (Excellent)"
elif score >= 80:
    grade = "B (Good)"
elif score >= 70:
    grade = "C (Average)"
elif score >= 60:
    grade = "D (Below Average)"
else:
    grade = "F (Fail)"

print(f"Score: {score} → Grade: {grade}")
```

```python
# ── Logical operators: and / or / not ────────────────────────────────
x = 15   # test number

# 'and' – both conditions must be True
if x > 10 and x < 20:
    print(f"{x} is between 10 and 20")

# 'or' – at least one condition must be True
if x < 5 or x > 12:
    print(f"{x} is outside the range 5–12")

# 'not' – inverts the condition
if not x == 10:
    print(f"{x} is not equal to 10")
```

```python
# ── Checking divisibility (modulo + conditionals) ────────────────────
n = 36   # number to test

if n % 2 == 0:
    print(f"{n} is Even")
else:
    print(f"{n} is Odd")

if n % 3 == 0 and n % 5 == 0:
    print(f"{n} is divisible by both 3 and 5")
elif n % 3 == 0:
    print(f"{n} is divisible by 3 only")
elif n % 5 == 0:
    print(f"{n} is divisible by 5 only")
```

```python
# ── for loop with range ───────────────────────────────────────────────
# range(1, 6) generates: 1, 2, 3, 4, 5

print("Counting from 1 to 5:")

for i in range(1, 6):      # i takes values 1, 2, 3, 4, 5
    print(f"  i = {i}")
```

```python
# ── Sum of first n natural numbers: S = n(n+1)/2 ─────────────────────

n = 10
total = 0

for i in range(1, n + 1):
    total += i

# Verify using the mathematical formula
formula_result = n * (n + 1) // 2

print(f"Sum (loop)    : {total}")
print(f"Sum (formula) : {formula_result}")
```

```python
# ── Factorial: n! = 1 × 2 × 3 × ... × n ─────────────────────────────

n = 6
factorial = 1

for i in range(1, n + 1):
    factorial *= i
    print(f"  Step {i}: {i}! so far = {factorial}")

print(f"\n{n}! = {factorial}")
```

```python
# ── while loop: Fibonacci sequence ───────────────────────────────────

# Fibonacci: 0, 1, 1, 2, 3, 5, 8, ...

a, b = 0, 1
limit = 100

print("Fibonacci numbers below", limit, ":")

while a < limit:
    print(a, end="  ")
    a, b = b, a + b

print()
```

```python
# ── Nested loop: multiplication table ────────────────────────────────

print("Multiplication Table (1–5)\n")

print("   ", end="")
for col in range(1, 6):
    print(f"{col:4}", end="")
print()

print("─" * 25)

for row in range(1, 6):
    print(f"{row:2} |", end="")
    for col in range(1, 6):
        print(f"{row * col:4}", end="")
    print()
```

```python
# ── Defining and calling a simple function ───────────────────────────

def square(x):          # define function 'square' with parameter x
    result = x ** 2     # compute x squared
    return result       # return the computed value

# Calling the function
print(square(4))        # → 16
print(square(9))        # → 81
print(square(-3))       # → 9
```

```python
# ── Functions for geometric formulas ────────────────────────────────

import math   # import Python's math module for π

def circle_area(r):
    """Returns area of a circle given radius r: A = πr²"""
    return math.pi * r ** 2

def circle_circumference(r):
    """Returns circumference: C = 2πr"""
    return 2 * math.pi * r

def triangle_area(base, height):
    """Returns area of a triangle: A = ½ × base × height"""
    return 0.5 * base * height

# Test the functions
r = 5

print(f"Circle Area (r={r})         : {circle_area(r):.4f}")
print(f"Circle Circumference (r={r}): {circle_circumference(r):.4f}")
print(f"Triangle Area (b=6, h=4)    : {triangle_area(6, 4):.4f}")
```

```python
# ── Default parameter values ─────────────────────────────────────────

def power(base, exponent=2):
    """Raises base to the given exponent. Default: square."""
    return base ** exponent

print(power(3))        # 9
print(power(3, 3))     # 27
print(power(2, 10))    # 1024
```

```python
# ── Function returning multiple values ───────────────────────────────

def quadratic_roots(a, b, c):
    """
    Returns both roots of ax² + bx + c = 0 using the quadratic formula.
    Returns (None, None) if there are no real roots.
    """
    discriminant = b ** 2 - 4 * a * c

    if discriminant < 0:
        return None, None

    root1 = (-b + discriminant ** 0.5) / (2 * a)
    root2 = (-b - discriminant ** 0.5) / (2 * a)

    return root1, root2

# Solve: x² - 5x + 6 = 0
r1, r2 = quadratic_roots(1, -5, 6)
print(f"Roots of x² - 5x + 6 = 0 : x₁ = {r1}, x₂ = {r2}")

# Solve: x² + 1 = 0
r1, r2 = quadratic_roots(1, 0, 1)
print(f"Roots of x² + 1 = 0      : {r1}, {r2} (no real solution)")
```

```python
# ── Lambda (anonymous) functions – one-line math ─────────────────────

double = lambda x: x * 2
cube = lambda x: x ** 3
celsius_to_fahrenheit = lambda c: (c * 9 / 5) + 32

print(double(7))
print(cube(4))
print(celsius_to_fahrenheit(100))
```

```python
import math   # import the entire math module

# ── Mathematical constants ────────────────────────────────────────────
print("π =", math.pi)      # 3.141592653589793
print("e =", math.e)       # 2.718281828459045
print("τ =", math.tau)     # 6.283185307179586 = 2π
print("∞ =", math.inf)     # positive infinity
```

```python
# ── Powers and roots ─────────────────────────────────────────────────

print("√16     =", math.sqrt(16))
print("∛27     =", 27 ** (1 / 3))
print("2^8     =", math.pow(2, 8))
print("|−5|    =", math.fabs(-5))
```

```python
# ── Logarithms ───────────────────────────────────────────────────────

print("log₂(8)    =", math.log2(8))
print("log₁₀(100) =", math.log10(100))
print("ln(e)      =", math.log(math.e))
print("log₅(125)  =", math.log(125, 5))
```

```python
# ── Rounding functions ────────────────────────────────────────────────

x = 4.7

print("floor (4.7) =", math.floor(x))
print("ceil  (4.7) =", math.ceil(x))
print("trunc (4.7) =", math.trunc(x))
print("round (4.7) =", round(x))
```

```python
# ── Trigonometry (angles in RADIANS) ─────────────────────────────────

deg = 45
rad = math.radians(deg)

print(f"Angle: {deg}° = {rad:.6f} radians")
print(f"sin(45°) = {math.sin(rad):.6f}")
print(f"cos(45°) = {math.cos(rad):.6f}")
print(f"tan(45°) = {math.tan(rad):.6f}")

# Pythagorean identity: sin²θ + cos²θ = 1
identity = math.sin(rad) ** 2 + math.cos(rad) ** 2
print(f"sin²(45°) + cos²(45°) = {identity:.6f}")
```
